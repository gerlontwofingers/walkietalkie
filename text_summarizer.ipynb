```python
import tiktoken
import requests
import os
from urllib.parse import urlparse

class AdvancedDocumentProcessor:
    def __init__(self, chunk_size=800, chunk_overlap=100, model_name="gpt-3.5-turbo"):
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        self.encoder = tiktoken.encoding_for_model(model_name)
        self.chunks = []
        self.raw_text = ""
        self.source_type = None  # 'local' or 'url'
        self.source_path = ""  # original path/URL
    
    def load(self, source):
        """
        Load document from either local file path or URL
        
        Args:
            source (str): Local file path or URL
        """
        self.source_path = source
        
        # Check if it's a URL
        if source.startswith(('http://', 'https://')):
            return self._load_from_url(source)
        else:
            return self._load_from_file(source)
    
    def _load_from_file(self, file_path):
        """Load text file with encoding handling"""
        try:
            if not os.path.exists(file_path):
                print(f"Error: File not found: {file_path}")
                return False
            
            with open(file_path, 'r', encoding='utf-8') as file:
                self.raw_text = file.read()
            self.source_type = 'local'
            print(f"Successfully loaded local file: {file_path}")
            return True
        except UnicodeDecodeError:
            # Fallback for other encodings
            try:
                with open(file_path, 'r', encoding='latin-1') as file:
                    self.raw_text = file.read()
                self.source_type = 'local'
                print(f"Successfully loaded local file (latin-1 encoding): {file_path}")
                return True
            except Exception as e:
                print(f"Error loading file with fallback encoding: {e}")
                return False
        except Exception as e:
            print(f"Error loading file: {e}")
            return False
    
    def _load_from_url(self, url):
        """Load text content from URL"""
        try:
            headers = {
                'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
            }
            response = requests.get(url, headers=headers, timeout=30)
            response.raise_for_status()  # Raises an HTTPError for bad responses
            
            self.raw_text = response.text
            self.source_type = 'url'
            print(f"Successfully loaded URL: {url}")
            return True
            
        except requests.exceptions.RequestException as e:
            print(f"Error loading URL: {e}")
            return False
        except Exception as e:
            print(f"Unexpected error loading URL: {e}")
            return False
    
    def count_tokens(self, text):
        return len(self.encoder.encode(text))
    
    def smart_chunking(self, text=None):
        """Split text into semantically meaningful chunks using paragraph boundaries"""
        if text is None:
            text = self.raw_text
        
        if not text.strip():
            print("Warning: No text content to chunk")
            self.chunks = []
            return []
        
        # Normalize line endings and split by paragraphs (double newlines)
        text = text.replace('\r\n', '\n').replace('\r', '\n')
        paragraphs = [p.strip() for p in text.split('\n\n') if p.strip()]
        
        chunks = []
        current_chunk = ""
        current_tokens = 0
        
        for paragraph in paragraphs:
            para_tokens = self.count_tokens(paragraph)
            
            # If this single paragraph is too large, we need to split it further
            if para_tokens > self.chunk_size:
                # First, save current chunk if it has content
                if current_chunk:
                    chunks.append(current_chunk.strip())
                    current_chunk = ""
                    current_tokens = 0
                
                # Split the large paragraph into sentences or smaller chunks
                large_chunks = self._split_large_paragraph(paragraph)
                for small_chunk in large_chunks:
                    small_tokens = self.count_tokens(small_chunk)
                    if current_tokens + small_tokens > self.chunk_size and current_chunk:
                        chunks.append(current_chunk.strip())
                        # Create overlap
                        overlap_text = self._create_token_overlap(current_chunk)
                        current_chunk = overlap_text + " " + small_chunk
                        current_tokens = self.count_tokens(current_chunk)
                    else:
                        current_chunk += " " + small_chunk if current_chunk else small_chunk
                        current_tokens += small_tokens
            else:
                # Normal paragraph processing
                if current_tokens + para_tokens > self.chunk_size and current_chunk:
                    chunks.append(current_chunk.strip())
                    # Create overlap from the end of current chunk
                    overlap_text = self._create_token_overlap(current_chunk)
                    current_chunk = overlap_text + " " + paragraph
                    current_tokens = self.count_tokens(current_chunk)
                else:
                    current_chunk += " " + paragraph if current_chunk else paragraph
                    current_tokens += para_tokens
        
        # Don't forget the last chunk
        if current_chunk:
            chunks.append(current_chunk.strip())
        
        self.chunks = chunks
        return chunks
    
    def _split_large_paragraph(self, paragraph):
        """Split a very large paragraph into smaller chunks"""
        # First try splitting by sentences
        sentences = [s.strip() for s in paragraph.split('.') if s.strip()]
        
        if len(sentences) > 1:
            chunks = []
            current_chunk = ""
            for sentence in sentences:
                sentence_with_period = sentence + '.' if not sentence.endswith('.') else sentence
                if self.count_tokens(current_chunk + " " + sentence_with_period) <= self.chunk_size:
                    current_chunk += " " + sentence_with_period if current_chunk else sentence_with_period
                else:
                    if current_chunk:
                        chunks.append(current_chunk)
                    current_chunk = sentence_with_period
            if current_chunk:
                chunks.append(current_chunk)
            return chunks
        else:
            # Fallback: split by fixed token size
            tokens = self.encoder.encode(paragraph)
            return [self.encoder.decode(tokens[i:i+self.chunk_size]) 
                   for i in range(0, len(tokens), self.chunk_size)]
    
    def _create_token_overlap(self, text):
        """Create overlap by taking the last N tokens from previous chunk"""
        tokens = self.encoder.encode(text)
        overlap_tokens = tokens[-self.chunk_overlap:]
        return self.encoder.decode(overlap_tokens)
    
    def get_chunks(self):
        return self.chunks
    
    def get_chunk_count(self):
        return len(self.chunks)
    
    def get_metadata(self):
        return {
            "source_type": self.source_type,
            "source_path": self.source_path,
            "total_chars": len(self.raw_text),
            "total_tokens": self.count_tokens(self.raw_text),
            "total_chunks": len(self.chunks),
            "avg_chunk_size": sum(len(c) for c in self.chunks) / len(self.chunks) if self.chunks else 0,
            "avg_chunk_tokens": sum(self.count_tokens(c) for c in self.chunks) / len(self.chunks) if self.chunks else 0
        }

    def preview_chunks(self, num_chunks=3):
        """Preview first few chunks for debugging"""
        preview = {}
        for i, chunk in enumerate(self.chunks[:num_chunks]):
            preview[f"chunk_{i+1}"] = {
                "tokens": self.count_tokens(chunk),
                "chars": len(chunk),
                "preview": chunk[:100] + "..." if len(chunk) > 100 else chunk
            }
        return preview
```

```python
# Example 1: Load from local file
processor = AdvancedDocumentProcessor(chunk_size=1000, chunk_overlap=150)
success = processor.load("/path/to/your/document.txt")
if success:
    chunks = processor.smart_chunking()
    print(f"Created {processor.get_chunk_count()} chunks from local file")
    print(f"Metadata: {processor.get_metadata()}")

# Example 2: Load from URL (like the King James Bible)
processor2 = AdvancedDocumentProcessor(chunk_size=800, chunk_overlap=100)
success = processor2.load("https://openbible.com/textfiles/kjv.txt")
if success:
    chunks = processor2.smart_chunking()
    print(f"Created {processor2.get_chunk_count()} chunks from URL")
    print(f"Metadata: {processor2.get_metadata()}")
    
    # Preview the first few chunks
    preview = processor2.preview_chunks(2)
    for chunk_id, info in preview.items():
        print(f"{chunk_id}: {info}")

# Example 3: Handle both sources in a unified way
sources = [
    # "/local/path/document.txt",
    "https://openbible.com/textfiles/kjv.txt"
]

for source in sources:
    processor = AdvancedDocumentProcessor()
    if processor.load(source):
        chunks = processor.smart_chunking()
        meta = processor.get_metadata()
        print(f"Processed {source}: {meta['total_chunks']} chunks, {meta['total_tokens']} tokens")
```
