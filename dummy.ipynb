```python
from mutagen.mp4 import MP4

def get_mp4_metadata(file_path):
    """
    Retrieve metadata from an MP4 file.

    Args:
        file_path (str): The path to the MP4 file.

    Returns:
        dict: A dictionary containing the metadata.

    Raises:
        FileNotFoundError: If the file does not exist.
    """
    try:
        mp4 = MP4(file_path)
        metadata = {
            'Title': mp4.get('title', [''])[0],
            'Artist': mp4.get('artist', [''])[0],
            'Album': mp4.get('album', [''])[0],
            'Genre': mp4.get('genre', [''])[0],
            'Duration': mp4.info.length,
            'Bitrate': mp4.info.bitrate,
            'Sample Rate': mp4.info.sample_rate,
            'Channels': mp4.info.channels
        }
        return metadata
    except FileNotFoundError:
        raise

# Example usage
file_path = '/Users/bagheera/Pictures/test/test/MVI_9897.MP4'
metadata = get_mp4_metadata(file_path)

# Display metadata in markdown format
for key, value in metadata.items():
    print(f'{key}: {value}')
```

```python
from PIL import Image
def get_date_taken(path, as_datetime: bool = True):
    if exif := Image.open(path)._getexif():
        return (
            datetime.strptime(exif[36867], '%Y:%m:%d %H:%M:%S')
            if as_datetime
            else exif[36867]
        )
    else:
        raise Exception('Image {0} does not have EXIF data.'.format(path))


get_date_taken("/Users/bagheera/Pictures/test/test/brotest/2025/winter/20250319/IMG_0808.JPG")
```

```python
# import datetime

def get_exif_creation_dates_video(path):
    EXIFTOOL_DATE_TAG_VIDEOS = "Create Date"
    EXIF_DATE_FORMAT = "%Y-%m-%d %H:%M:%S"
    EXIF_DATE_FORMATS = [
         "%Y-%m-%d %H:%M:%S",
         "%Y:%m:%d %H:%M:%S",
    ]
    # absolute_path = os.path.join( os.getcwd(), path )
    absolute_path = path

    process = subprocess.Popen(["exiftool", absolute_path], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    out, err = process.communicate()
    lines = out.decode("utf-8").split("\n")
    for l in lines:
        # print(l)
        if EXIFTOOL_DATE_TAG_VIDEOS in str(l):
            datetime_str = str(l.split(" : ")[1].strip())
            creation_dates = []
            creation_date = None
            for date_format in EXIF_DATE_FORMATS:
                try:
                    print(f'Reading vido metadata. Attempting to parse using format {date_format}...')
                    creation_date = datetime.datetime.strptime(datetime_str, date_format)
                    creation_dates += [creation_date]
                    print('...success!')
                except Exception:
                    print(f"Date format {date_format} was rejected. Trying new format.")
            if not creation_date:
                # pass
                 raise LookupError("Unable to parse date format in video.")

            # print(creation_date) #you will get 3 dates: Create Date, Track Create Date and Media Create Date 
            return min(creation_dates)

get_exif_creation_dates_video('/Users/bagheera/Pictures/test/test/MVI_9897.MP4')
```
